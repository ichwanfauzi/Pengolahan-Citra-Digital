```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = 'foto_diri.jpeg'
img = cv2.imread(image_path)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Gray
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Memberikan efek blur karena background sangat ramai
blurred = cv2.GaussianBlur(gray, (7, 7), 0)

# Canny Edge Detection
edges = cv2.Canny(blurred, threshold1=40, threshold2=120)

# Proses Contours Detection
contours, hierarchy = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
img_contours = img_rgb.copy()
# Menggambar kontur dengan warna hijau cerah
cv2.drawContours(img_contours, contours, -1, (0, 255, 0), 2)

plt.figure(figsize=(18, 6))

plt.subplot(1, 3, 1)
plt.imshow(img_rgb)
plt.title('Original Image')
plt.axis('on') 

plt.subplot(1, 3, 2)
plt.imshow(edges, cmap='gray')
plt.title('Canny Edge Detection')
plt.axis('on')

plt.subplot(1, 3, 3)
plt.imshow(img_contours)
plt.title('Contours Detection')
plt.axis('on')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = 'buah_daun.jpeg' 
img = cv2.imread(image_path)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Konversi ke HSV
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Proses Masking dan Segmentasi BUAH
lower_red1 = np.array([0, 100, 50])
upper_red1 = np.array([10, 255, 255])
mask1 = cv2.inRange(hsv, lower_red1, upper_red1)

lower_red2 = np.array([160, 100, 50])
upper_red2 = np.array([180, 255, 255])
mask2 = cv2.inRange(hsv, lower_red2, upper_red2)

# Menggabungkan kedua mask merah
mask_buah = mask1 + mask2

# Segmentasi Buah 
segmentasi_buah = cv2.bitwise_and(img_rgb, img_rgb, mask=mask_buah)

# Proses Masking dan Segmentasi DAUN (Hijau)
lower_green = np.array([30, 40, 40])
upper_green = np.array([85, 255, 255])
mask_daun = cv2.inRange(hsv, lower_green, upper_green)

# Segmentasi Daun
segmentasi_daun = cv2.bitwise_and(img_rgb, img_rgb, mask=mask_daun)

plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
plt.imshow(img_rgb)
plt.title('Gambar Asli')
plt.axis('on')

plt.subplot(2, 2, 2)
plt.imshow(mask_buah, cmap='gray')
plt.title('Mask Rambutan')
plt.axis('on')

plt.subplot(2, 2, 3)
plt.imshow(segmentasi_buah)
plt.title('Segmentasi Rambutan')
plt.axis('on')

plt.subplot(2, 2, 4)
plt.imshow(segmentasi_daun)
plt.title('Segmentasi Daun')
plt.axis('on')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = 'daun.jpeg'
img = cv2.imread(image_path)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Citra Biner (Adaptive Threshold)
biner_adaptive = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, 
                                       cv2.THRESH_BINARY_INV, 15, 4)
# 3. Setup Kernel 
kernel_3x3 = np.ones((3,3), np.uint8)
kernel_5x5 = np.ones((5,5), np.uint8)

# 4. Operasi Morfologi (Erosi dan Dilasi)
erosi_3x3 = cv2.erode(biner_adaptive, kernel_3x3, iterations=1)
dilasi_3x3 = cv2.dilate(biner_adaptive, kernel_3x3, iterations=1)
dilasi_5x5 = cv2.dilate(biner_adaptive, kernel_5x5, iterations=1)

plt.figure(figsize=(15, 10))
plt.suptitle('Hasil Operasi Morfologi - Daun (Grayscale + Adaptive)', fontsize=18, fontweight='bold', y=0.95)

plt.subplot(2, 3, 1), plt.imshow(img_rgb), plt.title('Asli', fontsize=14), plt.axis('off')
plt.subplot(2, 3, 2), plt.imshow(gray, cmap='gray'), plt.title('Grayscale', fontsize=14), plt.axis('off')
plt.subplot(2, 3, 3), plt.imshow(biner_adaptive, cmap='gray'), plt.title('Biner (Adaptive)', fontsize=14), plt.axis('off')

plt.subplot(2, 3, 4), plt.imshow(erosi_3x3, cmap='gray'), plt.title('Erosi (3x3)', fontsize=14), plt.axis('off')
plt.subplot(2, 3, 5), plt.imshow(dilasi_3x3, cmap='gray'), plt.title('Dilasi (3x3)', fontsize=14), plt.axis('off')
plt.subplot(2, 3, 6), plt.imshow(dilasi_5x5, cmap='gray'), plt.title('Dilasi (5x5)', fontsize=14), plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = 'tulisan_tangan.jpeg'
img = cv2.imread(image_path)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Citra Biner (Otsu Thresholding)
ret, biner_otsu = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

# Setup Kernel (3x3 dan 7x7)
kernel_3x3 = np.ones((3,3), np.uint8)
kernel_7x7 = np.ones((7,7), np.uint8)

# 4. Operasi Morfologi
opening_3x3 = cv2.morphologyEx(biner_otsu, cv2.MORPH_OPEN, kernel_3x3)
opening_7x7 = cv2.morphologyEx(biner_otsu, cv2.MORPH_OPEN, kernel_7x7)

closing_3x3 = cv2.morphologyEx(biner_otsu, cv2.MORPH_CLOSE, kernel_3x3)
closing_7x7 = cv2.morphologyEx(biner_otsu, cv2.MORPH_CLOSE, kernel_7x7)

plt.figure(figsize=(18, 9))
plt.suptitle('Hasil Operasi Morfologi - Tulisan Tangan', fontsize=18, fontweight='bold', y=0.95)

# Baris 1: Proses awal
plt.subplot(2, 4, 1), plt.imshow(img_rgb), plt.title('Asli', fontsize=14), plt.axis('off')
plt.subplot(2, 4, 2), plt.imshow(gray, cmap='gray'), plt.title('Grayscale', fontsize=14), plt.axis('off')
plt.subplot(2, 4, 3), plt.imshow(biner_otsu, cmap='gray'), plt.title('Biner (Otsu)', fontsize=14), plt.axis('off')

# Baris 2: Hasil Opening dan Closing (Perbandingan Kernel)
plt.subplot(2, 4, 5), plt.imshow(opening_3x3, cmap='gray'), plt.title('Opening (3x3)', fontsize=14), plt.axis('off')
plt.subplot(2, 4, 6), plt.imshow(opening_7x7, cmap='gray'), plt.title('Opening (7x7)', fontsize=14), plt.axis('off')
plt.subplot(2, 4, 7), plt.imshow(closing_3x3, cmap='gray'), plt.title('Closing (3x3)', fontsize=14), plt.axis('off')
plt.subplot(2, 4, 8), plt.imshow(closing_7x7, cmap='gray'), plt.title('Closing (7x7)', fontsize=14), plt.axis('off')

plt.subplots_adjust(wspace=0.1, hspace=0.2)
plt.show()
```
